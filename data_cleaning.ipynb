```python
import pandas as pd

```

```python
df = pd.read_csv('finn_raw.csv')

df.head()
df.info()
```

```python
df.isnull().sum()
```

### Henter in null values 

```python
df = df.dropna(subset='address')
df['total_price'] = df['total_price'].fillna(df['asking_price'] + df['fellesgjeld'])

print(df[df['bedrooms'].isna()])
df['bedrooms'] = df['bedrooms'].fillna(0)

print(df[df['rooms'].isna()])
df['rooms'] = df['rooms'].fillna(df['bedrooms'] + 1)

df = df.drop(columns=['BRA_i'])
```

```python
print(df[df['BRA_e'].isna()])
df['BRA_e'] = df['BRA_e'].fillna(0)

df.isnull().sum()

df['TBA'] = df['TBA'].fillna(df['m2'] + df['BRA_e'])


print(df[df['byggeår'].isna()])
df['byggeår'] = df['byggeår'].fillna(2027)

df.isnull().sum()

```

### Removing unnecessary features 

```python
print(df.columns)
df = df.drop(columns=['fellesgjeld', 'asking_price', 'url'])
df.head()
```

### Removing duplicates

```python
print(df.shape)
print('Number of duplicate rows:', df.duplicated().sum())
df[df.duplicated(keep=False)]
```

```python
df = df.drop_duplicates()
df.shape

```

#### Changing  name on columns and cells 

```python
df["total_price"] = df["total_price"].astype(int)
df["rooms"] = df["rooms"].astype(int)
df["bedrooms"] = df["bedrooms"].astype(int)
df["TBA"] = df["TBA"].astype(int)
df["BRA_e"] = df["BRA_e"].astype(int)
df["byggeår"] = df["byggeår"].astype(int)

df.head()
```

```python
df["address"] = (
    df["address"]
    .str.split(r"[,\-\d]", n=1, regex=True)
    .str[0]
    .str.strip()
)
df
```

```python
df = df.rename(columns={
    "address": "gatenavn",
    "m2": "internt_bruksareal",
    "TBA": "bruksareal",
    "BRA_e": "terrasseareal",
    "apartment_type": "leilighetstype",
    "total_price": "totalpris",
    "rooms": "rom",
    "bedrooms": "soverom",
    "byggeår": "byggeår"
})

df
```

```python
df = df[['totalpris','internt_bruksareal', 'terrasseareal', 'rom','byggeår', 'gatenavn']]
df.to_csv("finn_data_cleaned.csv", index=False)
```
